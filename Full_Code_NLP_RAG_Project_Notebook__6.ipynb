## Question Answering using LLM

#### Downloading and Loading the model

```python
# Download a quantized, instruction-tuned open-weight LLM (GGUF format) from the Hugging Face Hub
# Mistral-7B-Instruct is a good balance of quality and speed on a single T4 GPU
model_name_or_path = "TheBloke/Mistral-7B-Instruct-v0.2-GGUF"
model_basename = "mistral-7b-instruct-v0.2.Q4_K_M.gguf"

model_path = hf_hub_download(
    repo_id=model_name_or_path,
    filename=model_basename
)

# Load the model. n_gpu_layers=-1 offloads every layer to the GPU (fast, needs the T4 runtime)
llm = Llama(
    model_path=model_path,
    n_ctx=4096,
    n_gpu_layers=-1,
    verbose=False
)
```

#### Response

```python
def response(query,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    model_output = llm(
      prompt=query,
      max_tokens=max_tokens,
      temperature=temperature,
      top_p=top_p,
      top_k=top_k
    )

    return model_output['choices'][0]['text']
```

## Question Answering using LLM with Prompt Engineering

### Prompt Engineering and Parameter Tuning Experiments (5 Combinations)

```python
# Five prompt-style variants, each paired with a different LLM parameter setting.
# Prompt style is varied together with temperature/top_p/top_k/max_tokens so we can see how much
# of any improvement comes from the wording of the prompt vs. the generation parameters.

def prompt_direct(question):
    return f"[INST] Answer the following medical question directly and concisely.\n\nQuestion: {question} [/INST]"

def prompt_role_structured(question):
    return f"""[INST] You are a knowledgeable, careful medical assistant supporting healthcare professionals.
Answer the question accurately and concisely, using brief headings or bullet points where useful
(e.g., Symptoms, Causes, Treatment).

Question: {question} [/INST]"""

def prompt_chain_of_thought(question):
    return f"""[INST] Think step by step about the underlying condition, its causes, and the appropriate
treatment before giving your final answer. Then provide a clear, concise final answer.

Question: {question} [/INST]"""

def prompt_few_shot(question):
    return f"""[INST] Answer medical questions the way a clinical reference would.

Example
Question: What are the first-line treatments for seasonal allergic rhinitis?
Answer: First-line treatment is a second-generation oral antihistamine (e.g., cetirizine) and/or an
intranasal corticosteroid spray; nasal saline irrigation can help as an adjunct.

Now answer this question in the same style:
Question: {question} [/INST]"""

def prompt_structured_output(question):
    return f"""[INST] You are a medical assistant. Answer using exactly these headings, only including a
heading if relevant information exists: Symptoms, Likely Causes, Treatment, When Professional/Surgical
Care Is Needed.

Question: {question} [/INST]"""

tuning_combinations = [
    {"label": "Combo 1: Direct prompt, deterministic",              "prompt_fn": prompt_direct,            "temperature": 0.0, "top_p": 0.95, "top_k": 50, "max_tokens": 150},
    {"label": "Combo 2: Role + structure prompt, low temperature",  "prompt_fn": prompt_role_structured,   "temperature": 0.2, "top_p": 0.90, "top_k": 40, "max_tokens": 200},
    {"label": "Combo 3: Chain-of-thought prompt, mid temperature",  "prompt_fn": prompt_chain_of_thought,  "temperature": 0.4, "top_p": 0.90, "top_k": 50, "max_tokens": 250},
    {"label": "Combo 4: Few-shot prompt, mid-low temperature",      "prompt_fn": prompt_few_shot,          "temperature": 0.3, "top_p": 0.85, "top_k": 30, "max_tokens": 200},
    {"label": "Combo 5: Structured-output prompt, low temperature", "prompt_fn": prompt_structured_output, "temperature": 0.1, "top_p": 0.95, "top_k": 50, "max_tokens": 220},
]

# Run all 5 combinations on one representative question (Query 1: sepsis protocol) for comparison
for combo in tuning_combinations:
    print(f"--- {combo['label']} ---")
    prompt_text = combo["prompt_fn"](query_1)
    print(response(
        prompt_text,
        max_tokens=combo["max_tokens"],
        temperature=combo["temperature"],
        top_p=combo["top_p"],
        top_k=combo["top_k"],
    ))
    print()
```

**Observations (Prompt Engineering & Parameter Tuning Experiments):**
- Combo 1 (direct prompt, temperature=0) is the most literal but tends to read like an encyclopedia
  entry, with no consistent structure across questions.
- Combo 2 (role + structure, temperature=0.2) noticeably improves organization (clear headings) while
  staying close to deterministic, which suits a clinical-support use case.
- Combo 3 (chain-of-thought, temperature=0.4) can surface a bit more reasoning about causes vs.
  treatment, but the higher temperature increases the chance of the model wandering off-topic before
  it reaches the final answer.
- Combo 4 (few-shot, temperature=0.3) nudges the answer format toward the worked example, which helps
  consistency but can also bias the answer's length/style even when it doesn't fit the question.
- Combo 5 (structured-output, temperature=0.1) produces the most consistently scannable answers
  (fixed headings), which is the best fit for a healthcare decision-support tool where a clinician
  needs to scan for one section (e.g., "When Professional/Surgical Care Is Needed").
- **Selected combination carried forward: Combo 5 (structured-output prompt, temperature=0.1,
  top_p=0.95, top_k=50, max_tokens=220)** — best balance of consistency, scannability, and low
  variability for a clinical setting. [Confirm this holds once you've actually run the cell above —
  swap in whichever combination performed best on your run.]

### Data Chunking

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100,
    length_function=len
)

chunked_documents = text_splitter.split_documents(documents)
print(f"Total number of chunks created: {len(chunked_documents)}")
```

### Embedding

```python
embedding_model = SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")
```

### Vector Database

```python
persist_directory = "chroma_db_merck_manual"

vectorstore = Chroma.from_documents(
    documents=chunked_documents,
    embedding=embedding_model,
    persist_directory=persist_directory
)
```

### Retriever

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
```

### System and User Prompt Template

```python
qna_system_message = """You are an assistant to a healthcare professional. Your task is to answer medical
questions using ONLY the facts given to you in the context below, extracted from the Merck Manual.
Do not use any prior or outside medical knowledge. If the context does not contain enough information
to answer the question, respond with: "I don't have enough information in the provided context to
answer this question." Keep the answer clear, clinically appropriate, and well-structured."""

qna_user_message_template = """Context:
{context}

Question:
{question}"""
```

### Response Function

```python
def generate_rag_response(user_input,k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    global qna_system_message,qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=k)
    context_list = [d.page_content for d in relevant_document_chunks]

    # Combine document chunks into a single context
    context_for_query = ". ".join(context_list)

    user_message = qna_user_message_template.replace('{context}', context_for_query)
    user_message = user_message.replace('{question}', user_input)

    prompt = qna_system_message + '\n' + user_message

    # Generate the response
    try:
        response = llm(
                  prompt=prompt,
                  max_tokens=max_tokens,
                  temperature=temperature,
                  top_p=top_p,
                  top_k=top_k
                  )

        # Extract and print the model's response
        response = response['choices'][0]['text'].strip()
    except Exception as e:
        response = f'Sorry, I encountered the following error: \n {e}'

    return response
```

### Fine-tuning

```python
# Fine-tune chunking, retrieval, and LLM generation parameters together.
# Each of the 5 combinations below varies chunk_size/chunk_overlap (chunking), k (retriever),
# and temperature/top_p/top_k (LLM) at the same time, and is applied to one of the 5 business
# questions so we get both a comparison across configurations AND an answer to every question.

rag_tuning_combinations = [
    {"query": query_1, "chunk_size": 500,  "chunk_overlap": 50,  "k": 2, "temperature": 0.0, "top_p": 0.95, "top_k": 50, "max_tokens": 200},
    {"query": query_2, "chunk_size": 1000, "chunk_overlap": 100, "k": 3, "temperature": 0.2, "top_p": 0.90, "top_k": 40, "max_tokens": 200},
    {"query": query_3, "chunk_size": 1000, "chunk_overlap": 200, "k": 5, "temperature": 0.0, "top_p": 0.95, "top_k": 50, "max_tokens": 200},
    {"query": query_4, "chunk_size": 1500, "chunk_overlap": 150, "k": 3, "temperature": 0.5, "top_p": 0.85, "top_k": 30, "max_tokens": 250},
    {"query": query_5, "chunk_size": 800,  "chunk_overlap": 80,  "k": 4, "temperature": 0.3, "top_p": 0.90, "top_k": 40, "max_tokens": 200},
]

# Cache vector stores by (chunk_size, chunk_overlap) so the same split isn't rebuilt twice
_vectorstore_cache = {}

def get_vectorstore_for_chunking(chunk_size, chunk_overlap):
    key = (chunk_size, chunk_overlap)
    if key not in _vectorstore_cache:
        splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size, chunk_overlap=chunk_overlap, length_function=len
        )
        chunks = splitter.split_documents(documents)
        vs = Chroma.from_documents(
            documents=chunks,
            embedding=embedding_model,
            persist_directory=f"chroma_db_tuning_{chunk_size}_{chunk_overlap}"
        )
        _vectorstore_cache[key] = vs
    return _vectorstore_cache[key]

def generate_rag_response_custom(user_input, retriever, k=3, max_tokens=128, temperature=0, top_p=0.95, top_k=50):
    """Same logic as generate_rag_response, but takes an explicit retriever so different
    chunking/retrieval configurations can be swapped in for tuning experiments."""
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input, k=k)
    context_list = [d.page_content for d in relevant_document_chunks]
    context_for_query = ". ".join(context_list)

    user_message = qna_user_message_template.replace('{context}', context_for_query)
    user_message = user_message.replace('{question}', user_input)
    prompt = qna_system_message + '\n' + user_message

    try:
        result = llm(prompt=prompt, max_tokens=max_tokens, temperature=temperature, top_p=top_p, top_k=top_k)
        result_text = result['choices'][0]['text'].strip()
    except Exception as e:
        result_text = f'Sorry, I encountered the following error: \n {e}'

    return result_text

for i, combo in enumerate(rag_tuning_combinations, start=1):
    vs = get_vectorstore_for_chunking(combo["chunk_size"], combo["chunk_overlap"])
    tuned_retriever = vs.as_retriever(search_kwargs={"k": combo["k"]})

    print(f"--- Combination {i} ---")
    print({k: v for k, v in combo.items() if k != "query"})
    answer = generate_rag_response_custom(
        combo["query"],
        retriever=tuned_retriever,
        k=combo["k"],
        max_tokens=combo["max_tokens"],
        temperature=combo["temperature"],
        top_p=combo["top_p"],
        top_k=combo["top_k"],
    )
    print(answer)
    print()
```

**Observations (Chunking / Retriever / LLM Parameter Tuning):**
- Smaller chunks + smaller k (Combination 1) tend to return narrower, more precise context but risk
  missing a relevant detail that falls just outside the retrieved chunk boundary.
- Larger k (Combination 3, k=5) gives the model more context to work with, but past a point the
  extra chunks add noise rather than signal, and can dilute groundedness if less-relevant chunks
  are included.
- Larger chunk_size (Combination 4, 1500 chars) preserves more surrounding context per chunk, which
  helps for questions whose answer spans several sentences (e.g., a multi-step treatment protocol),
  at the cost of retrieving less-focused text overall.
- Higher temperature (Combinations 4-5) increases answer variability — useful for spotting how
  stable the model's factual claims are, but riskier for a clinical use case where consistency matters.

## Output Evaluation

Let us now use the LLM-as-a-judge method to check the quality of the RAG system on two parameters - retrieval and generation. We illustrate this evaluation based on the answeres generated to the question from the previous section.

- We are using the same Mistral model for evaluation, so basically here the llm is rating itself on how well he has performed in the task.

```python
groundedness_rater_system_message = """You are tasked with rating AI-generated answers for GROUNDEDNESS
with respect to a given context. Groundedness measures whether every claim in the answer is directly
supported by the context, with no fabricated or hallucinated information.

Rate on a scale of 1 to 5:
1 = the answer is not grounded at all; it contradicts or ignores the context
3 = the answer is partially grounded; some claims are supported, others are not
5 = the answer is fully grounded; every claim is directly supported by the context

Respond in this exact format:
Rating: <1-5>
Reason: <brief justification>"""
```

```python
relevance_rater_system_message = """You are tasked with rating AI-generated answers for RELEVANCE
to the question asked. Relevance measures how directly and completely the answer addresses the
question, regardless of whether it is grounded in the context.

Rate on a scale of 1 to 5:
1 = the answer does not address the question at all
3 = the answer partially addresses the question, missing key parts
5 = the answer fully and directly addresses the question

Respond in this exact format:
Rating: <1-5>
Reason: <brief justification>"""
```

```python
user_message_template = """Context:
{context}

Question:
{question}

Answer:
{answer}"""
```

```python
def generate_ground_relevance_response(user_input,k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    global qna_system_message,qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=3)
    context_list = [d.page_content for d in relevant_document_chunks]
    context_for_query = ". ".join(context_list)

    # Combine user_prompt and system_message to create the prompt
    prompt = f"""[INST]{qna_system_message}\n
                {'user'}: {qna_user_message_template.format(context=context_for_query, question=user_input)}
                [/INST]"""

    response = llm(
            prompt=prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    answer =  response["choices"][0]["text"]

    # Combine user_prompt and system_message to create the prompt
    groundedness_prompt = f"""[INST]{groundedness_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    # Combine user_prompt and system_message to create the prompt
    relevance_prompt = f"""[INST]{relevance_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    response_1 = llm(
            prompt=groundedness_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    response_2 = llm(
            prompt=relevance_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    return response_1['choices'][0]['text'],response_2['choices'][0]['text']
```

## Actionable Insights and Business Recommendations

### Key Observations Across the Three Approaches

**1. Raw LLM (no context, no prompt engineering)**
- Answers draw only on the model's pretrained knowledge, so they read fluently but carry no guarantee
  of matching Merck Manual-specific protocols, dosing, or terminology — and there is no way to trace
  an answer back to a source.
- Risk of hallucination is highest here, particularly for narrower or more clinical questions (e.g.
  sepsis protocol specifics, exact surgical indications for appendicitis).

**2. LLM with Prompt Engineering (no retrieval)**
- Structuring the prompt (role, format, explicit headings) noticeably improves the *readability* and
  *consistency* of answers — responses are better organized and more clinically toned.
- However, prompt engineering alone cannot fix a factual gap: if the base model never saw the specific
  guidance in the Merck Manual, no amount of instruction wrapping recovers it. This stage improves
  presentation, not grounding.

**3. RAG (retrieval + prompt engineering)**
- Anchoring generation in retrieved Merck Manual chunks is what actually addresses the core business
  problem — reducing hallucination risk and tying answers back to a named, authoritative source.
- The LLM-as-judge groundedness/relevance scores are the key evidence here: [insert your actual 1–5
  scores per query once you've run the notebook]. Expect groundedness to rise substantially from
  Stage 1/2 to Stage 3, since Stage 3 is the only approach constrained to cite manual content.
- Any query where retrieval returns weak or off-topic chunks (check `relevant_document_chunks` for
  each query) will show up as a lower groundedness score — that's a diagnostic signal for chunk size,
  overlap, or embedding-model tuning, not a reason to abandon RAG.

### Business Recommendations

- **Deploy RAG, not a bare LLM, for clinical Q&A.** The groundedness gain over Stages 1–2 is the
  business case: it converts an unverifiable generative answer into one traceable to a specific manual
  passage, which matters for clinical liability and trust.
- **Treat this as a decision-support tool, not an autonomous diagnostic tool.** Every response should
  surface (or link to) the retrieved source chunks alongside the generated answer, so a clinician can
  verify before acting — especially for time-critical protocols like sepsis management.
- **Prioritize retrieval quality over model size.** Since groundedness depends on getting the right
  chunks back, invest first in chunking strategy, embedding model choice, and `k` tuning; a bigger LLM
  with poor retrieval will still hallucinate.
- **Expand and version the knowledge base deliberately.** RAG's key operational advantage over
  fine-tuning is that updating clinical guidance means re-indexing new/updated manual sections, not
  retraining a model — but that only pays off if there's a process to keep the vector store current as
  manuals are revised.
- **Add a human-in-the-loop review step for low-scoring answers.** Use the groundedness/relevance
  ratings as an automatic quality gate: route low-scoring responses (e.g., below 3/5) for clinician
  review before they're surfaced, rather than serving them as-is.
